#Extract spectral classes using regex expression

```python
classes = r"([OBAFGKM])"

for value in range(len(df['SpType'])):
  print(df['SpType'][value])
  match = re.search(classes,str(df['SpType'][value]))
  if match is not None:
    df["SpTypeClass"] = match.group()




display(df)
```
